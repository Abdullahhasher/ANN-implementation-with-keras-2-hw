```python
import re
from sklearn.preprocessing import StandardScaler
def extract_numeric(val):
    match = re.search(r'\d+', str(val))
    return int(match.group()) if match else np.nan
x_train_clean = x_train.copy()
x_test_clean = x_test.copy()
x_train_clean['capture_rate'] = x_train_clean['capture_rate'].apply(extract_numeric)
x_test_clean['capture_rate'] = x_test_clean['capture_rate'].apply(extract_numeric)
numeric_cols = x_train_clean.select_dtypes(include=[np.number]).columns
x_train_numeric = x_train_clean[numeric_cols]
x_test_numeric = x_test_clean[numeric_cols]
x_train_numeric = x_train_numeric.fillna(x_train_numeric.median())
x_test_numeric = x_test_numeric.fillna(x_train_numeric.median())

sc = StandardScaler()
x_train_scaled = sc.fit_transform(x_train_numeric)
x_test_scaled = sc.transform(x_test_numeric)

x_train_scaled
```
